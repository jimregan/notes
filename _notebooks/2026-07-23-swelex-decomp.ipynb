```python
import json

decomp = []

with open("svlex.json") as f:
    for line in f.readlines():
        data = json.loads(line)
        orth = data.get("orthography", None)
        if orth is not None:
            decomposition = data.get("decomp", None)
            if decomposition is not None and isinstance(decomposition, list) and len(decomposition) > 1:
                decomp.append((orth, decomposition))
```
